```python

print("Mohan K V")
print("1BM25CS486")

rooms = {
    'A': 'Dirty',
    'B': 'Dirty'
}

# Starting position
vacuum_position = 'A'

# Display environment
def display():
    print("\nCurrent Room Status:")
    for room, status in rooms.items():
        print(room, ":", status)
    print("Vacuum Cleaner Position:", vacuum_position)


# Vacuum Cleaner Agent
def vacuum_agent():
    global vacuum_position

    print("\nVacuum Cleaner starts cleaning...")

    while True:

        # Check current room
        if rooms[vacuum_position] == 'Dirty':
            print("Room", vacuum_position, "is Dirty.")
            print("Cleaning Room", vacuum_position, "...")
            rooms[vacuum_position] = 'Clean'
            print("Room", vacuum_position, "is now Clean.")

        else:
            print("Room", vacuum_position, "is already Clean.")

        display()

        # Move to the next room
        if vacuum_position == 'A':
            vacuum_position = 'B'
            print("\nMoving from Room A to Room B")

        else:
            print("\nMoving from Room B to Room A")

            # Stop if all rooms are clean
            if all(status == 'Clean' for status in rooms.values()):
                print("\nAll rooms are clean!")
                break


# Main program
print("===== VACUUM CLEANER PROBLEM =====")

display()

vacuum_agent()

print("\nVacuum Cleaner Task Completed!")
```
