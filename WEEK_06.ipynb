```python

from itertools import product

def KB(P, Q, R):
    return (not Q or P) and (not P or not Q) and (Q or R)


def TT_Entails(query):
    symbols = [True, False]

    print("\nP\tQ\tR\tKB\tQuery")

    for P, Q, R in product(symbols, repeat=3):

        kb_result = KB(P, Q, R)
        query_result = query(P, Q, R)

        print(P, "\t", Q, "\t", R, "\t",
              kb_result, "\t", query_result)

        if kb_result and not query_result:
            return False

    return True



print("PROPOSITIONAL LOGIC")

print("\n1. Does KB entail R?")
result = TT_Entails(lambda P, Q, R: R)
print("KB entails R:", result)


print("\n2. Does KB entail R -> P?")
result = TT_Entails(lambda P, Q, R: not R or P)
print("KB entails R -> P:", result)


print("\n3. Does KB entail Q -> R?")
result = TT_Entails(lambda P, Q, R: not Q or R)
print("KB entails Q -> R:", result)
```
